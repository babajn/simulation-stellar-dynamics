# Forward Euler Method Explanation

### Objective
The **Forward Euler Method** is a simple and widely used numerical technique for solving **ordinary differential equations (ODEs)**. It approximates the solution to an initial value problem by iterating from an initial point using a small time step. This method is first-order accurate and easy to implement but is less stable compared to higher-order methods.

### Key Equation

The general form of an initial value problem is:

$$\frac{dy}{dt} = f(t, y), \quad y(t_0) = y_0$$

Here, $y(t)$ is the function we are trying to approximate, $t$ is the independent variable (often representing time), and $f(t, y)$ represents the derivative of $y$ with respect to $t$.

The **Forward Euler Method** uses the following formula to estimate $y$ at each step:

$$
y_{n+1} = y_n + h \cdot f(t_n, y_n)
$$

Where:
- $y_{n+1}$ is the value of $y$ at the next step.
- $y_n$ is the value of $y$ at the current step.
- $h$ is the step size (the difference between two consecutive $t$-values).
- $f(t_n, y_n)$ is the slope of $y$ at the current step.

### Step-by-Step Procedure

1. **Initialization**:
   - Start with an initial condition: $y(t_0) = y_0$.
   - Set the step size $h$ and the range of time over which you want to calculate the solution.

2. **Iteration**:
   - For each time step, compute the next value $y_{n+1}$ using the formula:
   $$
   y_{n+1} = y_n + h \cdot f(t_n, y_n)
   $$
   - Update $t_n$ to $t_{n+1} = t_n + h$.

3. **Termination**:
   - Repeat the iteration until the desired time range is covered.

### Advantages
- **Simple and easy to implement**: The method requires only basic arithmetic operations and is intuitive to use.
- **Computationally efficient**: It performs fewer calculations compared to higher-order methods like Runge-Kutta.

### Disadvantages
- **Stability issues**: For stiff ODEs or large step sizes, the method can become unstable, leading to incorrect solutions.
- **Low accuracy**: The method is only first-order accurate, meaning the error reduces linearly with the step size.

### Summary
The **Forward Euler Method** is a straightforward and computationally efficient method for solving ODEs. However, due to its stability limitations and lower accuracy, it is often replaced by more sophisticated methods (e.g., Runge-Kutta) in applications requiring high precision or large step sizes.

## Explanation of the `euler_method` Function

This function implements the **explicit Forward Euler method** for solving ordinary differential equations (ODEs) of the form $ \frac{dy}{dt} = f(t, y) $. The Forward Euler method is an **explicit** method, meaning it uses the current information at each time step to compute the next value directly.

### Function Parameters:
- **`f` (callable)**: The ODE function representing $ f(t, y, \alpha) $, where $ \alpha $ is a parameter of the ODE.
- **`y0` (float)**: The initial condition, representing $ y(t_0) = y_0 $.
- **`t0` (float)**: The starting time.
- **`t_end` (float)**: The end time of the simulation.
- **`h` (float)**: The step size for time discretization.
- **`alpha` (float)**: A parameter of the ODE that influences the behavior of $ f $.

### Returns:
- **`t_values` (numpy.ndarray)**: An array containing the time points from $ t_0 $ to $ t_{\text{end}} $.
- **`y_values` (numpy.ndarray)**: An array containing the numerical solution for $ y $ at each time point.

### Key Steps in the Code:

1. **Time and Solution Arrays Initialization**:
   - The number of steps $ N $ is calculated based on the total simulation time and the step size.
   - The `t_values` array holds the time points, and `y_values` stores the corresponding $ y $-values.
   - The initial condition $ y_0 $ is assigned to `y_values[0]`.

2. **Main Loop for Time Stepping**:
   - For each time step $ n $, the current time $ t_n $ and the corresponding solution $ y_n $ are retrieved.
   - The next value $ y_{\text{new}} $ is computed using the Forward Euler formula:
     $$
     y_{\text{new}} = y_n + h \cdot f(t_n, y_n, \alpha)
     $$
     This formula updates the solution at the next time step by using the current derivative information from $ f(t_n, y_n, \alpha) $.

3. **Update the Solution**:
   - Once $ y_{\text{new}} $ is computed, it is stored in `y_values[n+1]`.

4. **Final Output**:
   - The function returns the `t_values` (time points) and `y_values` (corresponding solutions at each time point).

### Example Workflow:
1. **Time stepping**: The solution progresses step-by-step from $ t_0 $ to $ t_{\text{end}} $, updating the solution using the Forward Euler method formula.
2. **Explicit Update**: At each time step, the current solution $ y_n $ and the function value $ f(t_n, y_n, \alpha) $ are used to compute the next value $ y_{\text{new}} $.
3. **Simple and Direct**: The Forward Euler method is straightforward but can become unstable for stiff ODEs or when the step size $ h $ is too large.

### Key Advantages:
- **Simple and Easy to Implement**: The Forward Euler method is an explicit method, making it computationally cheap and easy to implement.
- **Direct Calculation**: The next solution value is calculated directly from the current information without needing to solve an implicit equation.
  
### Limitations:
- **Limited Stability**: The Forward Euler method may become unstable for large step sizes or for stiff ODEs, meaning that smaller step sizes are often required for accurate results.
- **First-Order Accuracy**: The method has first-order accuracy, meaning the error decreases linearly as the step size is reduced.

```python
# Define the Forward Euler Method Function
def euler_method(f, y0, t0, t_end, h, alpha):
    """
    Solve the ODE dy/dt = f(t, y) using the explicit Forward Euler method.

    Parameters:
        f (callable): Function representing dy/dt = f(t, y)
        y0 (float): Initial condition y(t0) = y0
        t0 (float): Initial time
        t_end (float): End time
        h (float): Step size
        alpha (float): Parameter in the ODE

    Returns:
        t_values (numpy.ndarray): Array of time points
        y_values (numpy.ndarray): Array of numerical solutions at each time point
    """
    if not (np.isfinite(h) and h > 0 and np.isfinite(t0) and np.isfinite(t_end) and t_end > t0):
        raise ValueError("Require finite t_end > t0 and h > 0")
    N = max(1, int(round((t_end - t0) / h)))
    h = (t_end - t0) / N  # Actual uniform step matches returned times
    t_values = np.linspace(t0, t_end, N + 1)
    y_values = np.zeros(N + 1)
    y_values[0] = y0

    for n in range(N):
        t_n = t_values[n]
        y_n = y_values[n]
        y_new = y_n + h * f(t_n, y_n, alpha)
        y_values[n + 1] = y_new

    return t_values, y_values
```

# Numerical Methods for Solving Linear Ordinary Differential Equations (ODEs)

```python
def f(t, y, alpha):
    """
    Defines the right-hand side of the differential equation dy/dt = -alpha * y.

    Parameters:
        t (float): The current time
        y (float): The current value of y
        alpha (float): The coefficient alpha in the equation

    Returns:
        float: The value of dy/dt
    """
    return -alpha * y

def analytical_solution(t, alpha):
    """
    Computes the analytical solution y(t) = e^(-alpha * t).

    Parameters:
        t (float or np.ndarray): The time at which to evaluate the solution
        alpha (float): The coefficient alpha in the equation

    Returns:
        float or np.ndarray: The analytical solution y(t)
    """
    return np.exp(-alpha * t)
```

  Compares the numerical solution obtained using the Forward Euler method with the analytical solution of the differential equation $\frac{dy}{dt} = -\alpha y$ for various step sizes.

- ## Process:
  1. **Initialization:**  
     Sets up the initial conditions and a range of step sizes to test.
  2. **Analytical Solution:**  
     Computes the exact solution at the final time for reference.
  3. **Numerical Simulation:**  
     For each step size, it:
     - Solves the ODE using the Euler method.
     - Measures the computation time.
     - Calculates the relative error between the numerical and analytical solutions.
  4. **Results Display:**  
     Prints a table of step sizes, numerical solutions, errors, and computation times.
  5. **Visualization:**  
     Plots the convergence behavior and computation time as functions of the step size.

- ##  Key Takeaways:
  - **Accuracy vs. Step Size:**  
    Smaller step sizes generally lead to more accurate numerical solutions but require more computation time.
  - **Euler Method Limitations:**  
    Being a first-order method, Euler’s method has significant errors for larger step sizes, highlighting the need for higher-order methods in practice.
  - **Performance Analysis:**  
    The code provides insights into how the choice of step size affects both the accuracy and the efficiency of the numerical method.

```python
# Main function to execute the comparison and analysis
def main():
    # Initial conditions and settings
    alpha = 5
    y0 = 1.0
    t0 = 0.0
    t_end = 1.0
    step_sizes = [#0.5,
                  0.1, 0.05, 0.025, 0.0125, 0.00625, 0.003125, 0.0015625,
                  0.00078125, 0.000390625, 0.0001953125, 0.0001]

    # Compute the analytical solution at t_end
    y_true = analytical_solution(t_end, alpha)
    print(f"Analytical solution y({t_end}) = {y_true:.10f}\n")

    # List to store errors and computation times for each step size
    errors = []
    times = []

    # Header for the output table
    print("Euler Method Numerical Solution, Errors, and Times:")
    print(f"{'Step Size h':>15} | {'Numerical y_N':>15} | {'Relative error':>15} | {'Time (s)':>10}")
    print("-" * 75)

    # Compute numerical solutions, errors, and times for each step size
    for h in step_sizes:
        start_time = time.perf_counter()  # Start timing

        # Get the entire list or array of solutions from euler_method
        t_values, y_numerical_all = euler_method(f, y0, t0, t_end, h, alpha)

        end_time = time.perf_counter()  # End timing
        elapsed_time = end_time - start_time  # Calculate elapsed time

        # Use the final value from the numerical solution
        y_numerical = y_numerical_all[-1]

        # Compute the relative error
        error = abs(y_numerical - y_true) / abs(y_true)
        errors.append(error)
        times.append(elapsed_time)

        # Print the results in scientific notation
        print(f"{h:15.5e} | {y_numerical:15.10e} | {error:15.10e} | {elapsed_time:10.5e}")

    # Plotting step size vs. error on a log-log scale
    size = 3
    plt.figure(figsize=(4 * size, 3 * size))
    plt.loglog(step_sizes, errors, 'o-', markersize=10, lw=4, label='Euler Method Error')
    plt.xlabel('Step Size h')
    plt.ylabel('Relative Error')
    plt.title('Convergence of Forward Euler Method')
    plt.legend()
    plt.grid(True, which="both", ls="--")
    plt.show()

    # Plotting step size vs. computation time on a log-log scale
    plt.figure(figsize=(4 * size, 3 * size))
    plt.loglog(step_sizes, times, 'o-', markersize=10, lw=4, label='Computation Time')
    plt.xlabel('Step Size h')
    plt.ylabel('Computation Time (s)')
    plt.title('Computation Time vs Step Size')
    plt.legend()
    plt.grid(True, which="both", ls="--")
    plt.show()

if __name__ == "__main__":
    main()
```

# Numerical Methods for Solving Nonlinear Ordinary Differential Equations (ODEs)

```python
def f_nl(t, y, alpha):
    """
    Defines the right-hand side of the differential equation dy/dt = -alpha * y^2.

    Parameters:
        t (float): The current time
        y (float): The current value of y
        alpha (float): The coefficient alpha in the equation

    Returns:
        float: The value of dy/dt
    """
    return -alpha * y**2

def df_nl_dy(t, y, alpha):
    """
    Defines the derivative of f_nl with respect to y.

    Parameters:
        t (float): The current time
        y (float): The current value of y
        alpha (float): The coefficient alpha in the equation

    Returns:
        float: The value of df/dy
    """
    return -2 * alpha * y

def analytical_solution_nl(t, alpha, y0=1.0):
    """
    Computes the analytical solution y(t) = y0 / (1 + alpha * y0 * t).

    Parameters:
        t (float or np.ndarray): The time(s) at which to evaluate the solution
        alpha (float): The coefficient alpha in the equation
        y0 (float): The initial condition y0

    Returns:
        float or np.ndarray: The analytical solution y(t)
    """
    return y0 / (1 + alpha * y0 * t)
```

# Numerical Solution of Nonlinear Ordinary Differential Equations Using the Backward Euler Method

## Explanation of the `backward_euler_method` Function

This function implements the **Backward Euler method** for numerically solving ordinary differential equations (ODEs) of the form $ \frac{dy}{dt} = f(t, y) $. The Backward Euler method is an implicit method, meaning it requires solving a nonlinear equation at each time step. In this implementation, **Newton-Raphson iteration** is used to solve the implicit equation for each step.

### Function Parameters:
- **`f` (callable)**: The ODE function representing $ f(t, y, \alpha) $, where $ \alpha $ is a parameter of the ODE.
- **`df_dy` (callable)**: The derivative of the ODE function with respect to $ y $, required for Newton-Raphson iterations.
- **`y0` (float)**: The initial condition, representing $ y(t_0) = y_0 $.
- **`t0` (float)**: The starting time.
- **`t_end` (float)**: The end time of the simulation.
- **`h` (float)**: The step size for the time discretization.
- **`alpha` (float)**: A parameter of the ODE that influences the behavior of $ f $.
- **`tol` (float)**: Tolerance for the convergence of the Newton-Raphson method (default is $ 10^{-10} $).
- **`max_iter` (int)**: Maximum number of iterations allowed for Newton-Raphson convergence.

### Returns:
- **`t_values` (numpy.ndarray)**: An array containing the time points from $ t_0 $ to $ t_{\text{end}} $.
- **`y_values` (numpy.ndarray)**: An array containing the numerical solution for $ y $ at each time point.

### Key Steps in the Code:

1. **Time and Solution Arrays Initialization**:
    - The number of steps $ N $ is calculated based on the total simulation time and step size.
    - The `t_values` array holds the time points, and `y_values` stores the corresponding $ y $-values.
    - The initial condition $ y_0 $ is assigned to `y_values[0]`.

2. **Main Loop for Time Stepping**:
    - For each time step $ n $, the current time $ t_n $ and solution $ y_n $ are updated.
    - The new time point $ t_{\text{new}} = t_n + h $ is calculated.

3. **Implicit Equation Setup**:
    - The implicit Backward Euler equation is expressed as:
      $$
      G(y_{\text{new}}) = y_{\text{new}} - y_n - h \cdot f(t_{\text{new}}, y_{\text{new}}, \alpha) = 0
      $$
      This equation needs to be solved iteratively to find $ y_{\text{new}} $.

4. **Newton-Raphson Method**:
    - The **Newton-Raphson iteration** is used to solve for $ y_{\text{new}} $. The iteration continues until the difference between successive guesses is less than the tolerance `tol`.
    - If the derivative $ G'(y_{\text{new}}) $ becomes zero, an error is raised to prevent division by zero.
    - If convergence is not achieved within the maximum number of iterations, a runtime error is raised.

5. **Update the Solution**:
    - Once $ y_{\text{new}} $ is computed, it is stored in `y_values[n+1]`.

### Example Workflow:
1. **Time stepping**: The solution progresses step-by-step from $ t_0 $ to $ t_{\text{end}} $, solving the implicit equation at each time step.
2. **Newton-Raphson**: At each time step, Newton-Raphson is used to solve the nonlinear equation resulting from the Backward Euler discretization.
3. **Convergence checks**: The solution checks for convergence using the provided tolerance `tol`.

### Key Advantages:
- **Stability**: The Backward Euler method is known for its stability, especially for stiff ODEs.
- **Implicit Method**: Unlike explicit methods, implicit methods like Backward Euler allow for larger step sizes without causing instability, at the cost of having to solve an equation at each step.

```python
def backward_euler_method(f, df_dy, y0, t0, t_end, h, alpha, tol=1e-10, max_iter=100):
    """
    Solve the ODE dy/dt = f(t, y) using the implicit Backward Euler method.

    Parameters:
        f (callable): Function representing dy/dt = f(t, y, alpha)
        df_dy (callable): Function representing the derivative df/dy = df/dy(t, y, alpha)
        y0 (float): Initial condition y(t0) = y0
        t0 (float): Initial time
        t_end (float): End time
        h (float): Step size
        alpha (float): Parameter in the ODE
        tol (float): Tolerance for Newton-Raphson convergence
        max_iter (int): Maximum iterations for Newton-Raphson

    Returns:
        t_values (numpy.ndarray): Array of time points
        y_values (numpy.ndarray): Array of numerical solutions at each time point
    """
    if not (np.isfinite(h) and h > 0 and np.isfinite(t0) and np.isfinite(t_end) and t_end > t0):
        raise ValueError("Require finite t_end > t0 and h > 0")
    N = max(1, int(round((t_end - t0) / h)))
    h = (t_end - t0) / N  # Actual uniform step matches returned times
    t_values = np.linspace(t0, t_end, N + 1)
    y_values = np.zeros(N + 1)
    y_values[0] = y0

    for n in range(N):
        t_n = t_values[n]
        y_n = y_values[n]
        t_new = t_n + h

        # Define the function G(y_new) = y_new - y_n - h * f(t_new, y_new, alpha) = 0
        def G(y_new):
            return y_new - y_n - h * f(t_new, y_new, alpha)

        # Define the derivative G'(y_new) = 1 - h * df_dy(t_new, y_new, alpha)
        def G_prime(y_new):
            return 1 - h * df_dy(t_new, y_new, alpha)

        # Initial guess: y_n (can be improved based on problem specifics)
        y_guess = y_n

        # Newton-Raphson Iteration
        for iteration in range(max_iter):
            G_val = G(y_guess)
            Gp_val = G_prime(y_guess)
            if Gp_val == 0:
                raise ZeroDivisionError(f"Derivative is zero at step {n}, h={h}. No solution found.")
            y_new = y_guess - G_val / Gp_val
            if abs(y_new - y_guess) < tol:
                break
            y_guess = y_new
        else:
            raise RuntimeError(f"Backward Euler method did not converge at step {n}, h={h}")

        y_values[n + 1] = y_new

    return t_values, y_values
```

## Explanation of the `backward_euler_method_fsolve` Function

This function solves a nonlinear ordinary differential equation (ODE) of the form $ \frac{dy}{dt} = f(t, y) $ using the **Backward Euler method**. The key difference here is that it leverages `scipy.optimize.fsolve` to solve the implicit equation at each time step.

### Function Parameters:
- **`f` (callable)**: The ODE function representing $ f(t, y, \alpha) $, where $ \alpha $ is a parameter of the ODE.
- **`y0` (float)**: The initial condition, representing $ y(t_0) = y_0 $.
- **`t0` (float)**: The starting time.
- **`t_end` (float)**: The end time of the simulation.
- **`h` (float)**: The step size for time discretization.
- **`alpha` (float)**: A parameter of the ODE that influences the behavior of $ f $.
- **`tol` (float)**: The tolerance for the convergence of `fsolve` (default is $ 10^{-10} $).

### Returns:
- **`t_values` (numpy.ndarray)**: An array containing the time points from $ t_0 $ to $ t_{\text{end}} $.
- **`y_values` (numpy.ndarray)**: An array containing the numerical solution for $ y $ at each time point.

### Key Steps in the Code:

1. **Time and Solution Arrays Initialization**:
   - The number of steps $ N $ is calculated based on the total simulation time and step size.
   - The `t_values` array holds the time points, and `y_values` stores the corresponding $ y $-values.
   - The initial condition $ y_0 $ is assigned to `y_values[0]`.

2. **Main Loop for Time Stepping**:
   - For each time step $ n $, the current time $ t_n $ and solution $ y_n $ are updated.
   - The new time point $ t_{\text{new}} = t_n + h $ is calculated.

3. **Implicit Equation Setup**:
   - The implicit Backward Euler equation is expressed as:
     $$
     G(y_{\text{new}}) = y_{\text{new}} - y_n - h \cdot f(t_{\text{new}}, y_{\text{new}}, \alpha) = 0
     $$
     This equation needs to be solved at each step to find $ y_{\text{new}} $.

4. **Solving the Implicit Equation with `fsolve`**:
   - The `fsolve` function from `scipy.optimize` is used to solve the nonlinear equation $ G(y_{\text{new}}) = 0 $. The initial guess for $ y_{\text{new}} $ is the previous solution $ y_n $.
   - The solution is found within the specified tolerance `tol`.

5. **Update the Solution**:
   - Once $ y_{\text{new}} $ is computed using `fsolve`, it is stored in `y_values[n+1]`.

### Example Workflow:
1. **Time stepping**: The solution progresses step-by-step from $ t_0 $ to $ t_{\text{end}} $, solving the implicit equation at each time step using `fsolve`.
2. **Implicit Equation Solving**: At each time step, the implicit Backward Euler equation is solved by the `fsolve` function to find the next value $ y_{\text{new}} $.
3. **Convergence**: The solution is computed to within the tolerance `tol` specified for the `fsolve` solver.

### Key Advantages:
- **Implicit Solver**: The use of `fsolve` provides a robust method for solving the implicit Backward Euler equations, allowing for the method to handle stiff ODEs effectively.
- **Stability**: The Backward Euler method is stable for large step sizes and stiff equations, making it suitable for a variety of ODEs where explicit methods may fail.

```python
from scipy.optimize import fsolve
def backward_euler_method_fsolve(f, y0, t0, t_end, h, alpha, tol=1e-10):
    """
    Solve the ODE dy/dt = f(t, y) using the implicit Backward Euler method with scipy.optimize.fsolve.

    Parameters:
        f (callable): Function representing dy/dt = f(t, y, alpha)
        y0 (float): Initial condition y(t0) = y0
        t0 (float): Initial time
        t_end (float): End time
        h (float): Step size
        alpha (float): Parameter in the ODE
        tol (float): Tolerance for solver convergence

    Returns:
        t_values (numpy.ndarray): Array of time points
        y_values (numpy.ndarray): Array of numerical solutions at each time point
    """
    if not (np.isfinite(h) and h > 0 and np.isfinite(t0) and np.isfinite(t_end) and t_end > t0):
        raise ValueError("Require finite t_end > t0 and h > 0")
    N = max(1, int(round((t_end - t0) / h)))
    h = (t_end - t0) / N  # Actual uniform step matches returned times
    t_values = np.linspace(t0, t_end, N + 1)
    y_values = np.zeros(N + 1)
    y_values[0] = y0

    for n in range(N):
        t_n = t_values[n]
        y_n = y_values[n]
        t_new = t_n + h

        # Define the implicit function G(y_new) = y_new - y_n - h * f(t_new, y_new, alpha)
        def G(y_new):
            return y_new - y_n - h * f(t_new, y_new, alpha)

        # Use scipy's fsolve function to solve G(y_new) = 0
        y_new, info, ier, message = fsolve(G, y_n, xtol=tol, full_output=True)
        if ier != 1 and np.max(np.abs(info["fvec"])) > tol:
            raise RuntimeError(message)

        # Update the solution at the next time step
        y_values[n + 1] = y_new[0]

    return t_values, y_values
```
